Create a Quiz Engine that dynamically evaluates participants.
The system should manage a question bank categorized by difficulty levels. During quiz
attempts, questions should appear in a pseudo-random order without using random
libraries.
Scoring must vary by difficulty, include negative marking, and generate detailed score
reports. String handling and nested data structures must be used extensivel

```python
question_bank = {
    "easy": [
        {
            "question": "What is the output of: print(2 + 3)?",
            "options": ["4", "5", "6", "23"],
            "answer": "5"
        },
        {
            "question": "Which keyword is used to define a function in Python?",
            "options": ["func", "define", "def", "function"],
            "answer": "def"
        }
    ],
    "medium": [
        {
            "question": "What does len('Python') return?",
            "options": ["5", "6", "7", "Error"],
            "answer": "6"
        },
        {
            "question": "Which data type is immutable?",
            "options": ["List", "Set", "Dictionary", "Tuple"],
            "answer": "Tuple"
        }
    ],
    "hard": [
        {
            "question": "What is the output of: bool('False')?",
            "options": ["False", "True", "Error", "None"],
            "answer": "True"
        }
    ]
}

# Scoring rules
score_rules = {
    "easy": {"correct": 2, "wrong": -1},
    "medium": {"correct": 4, "wrong": -2},
    "hard": {"correct": 6, "wrong": -3}
}

# Pseudo-random index generator (NO random module)
def pseudo_shuffle(items):
    shuffled = []
    seed = len(items) * 7
    for i in range(len(items)):
        index = (seed + i * 3) % len(items)
        shuffled.append(items[index])
    return shuffled

# Quiz execution
def start_quiz():
    total_score = 0
    report = []

    print("\n=== QUIZ STARTED ===\n")

    for level in question_bank:
        questions = pseudo_shuffle(question_bank[level])

        for q in questions:
            print(f"Difficulty: {level.upper()}")
            print(q["question"])

            for i, option in enumerate(q["options"], start=1):
                print(f"{i}. {option}")

            user_input = input("Your answer: ").strip()

            if user_input == q["answer"]:
                score = score_rules[level]["correct"]
                result = "Correct"
            else:
                score = score_rules[level]["wrong"]
                result = "Wrong"

            total_score += score

            report.append({
                "question": q["question"],
                "your_answer": user_input,
                "correct_answer": q["answer"],
                "result": result,
                "score_change": score
            })

            print(f"{result}! Score change: {score}\n")

    generate_report(report, total_score)

# Detailed score report
def generate_report(report, total_score):
    print("\n=== DETAILED SCORE REPORT ===\n")

    for item in report:
        print(f"Q: {item['question']}")
        print(f"Your Answer: {item['your_answer']}")
        print(f"Correct Answer: {item['correct_answer']}")
        print(f"Result: {item['result']}")
        print(f"Score Change: {item['score_change']}")
        print("-" * 40)

    print(f"\nFINAL SCORE: {total_score}")
    print("=== QUIZ ENDED ===")

start_quiz()
```

Design a University Course Registration System.
The system should allow creation of courses with credits and seat limits, student
registration, and course enrollment.
Constraints include maximum credit limits per student, prevention of duplicate
registrations, and seat availability enforcement. Summary reports for students and courses
must be generated.


```python

MAX_CREDITS_PER_STUDENT = 20


courses = {}
students = {}
def add_course(course_id, name, credits, seats):
    courses[course_id] = {
        "name": name,
        "credits": credits,
        "seats": seats,
        "enrolled": []
    }


def register_student(student_id, name):
    students[student_id] = {
        "name": name,
        "courses": [],
        "total_credits": 0
    }
def enroll_student(student_id, course_id):
    student = students[student_id]
    course = courses[course_id]


    if course_id in student["courses"]:
        print("Student already enrolled in this course.")
        return
    if len(course["enrolled"]) >= course["seats"]:
        print("No seats available.")
        return

    if student["total_credits"] + course["credits"] > MAX_CREDITS_PER_STUDENT:
        print("Credit limit exceeded.")
        return


    student["courses"].append(course_id)
    student["total_credits"] += course["credits"]
    course["enrolled"].append(student_id)

    print(f"Student {student['name']} enrolled in {course['name']}")


def student_report(student_id):
    student = students[student_id]
    print(f"\nStudent Name: {student['name']}")
    print("Enrolled Courses:")

    for cid in student["courses"]:
        print(f"- {courses[cid]['name']}")

    print(f"Total Credits: {student['total_credits']}")


def course_report(course_id):
    course = courses[course_id]
    print(f"\nCourse: {course['name']}")
    print(f"Credits: {course['credits']}")
    print(f"Seats: {course['seats']}")
    print("Enrolled Students:")

    for sid in course["enrolled"]:
        print(f"- {students[sid]['name']}")

add_course("C101", "Python Programming", 4, 2)
add_course("C102", "Data Structures", 5, 1)

register_student("S1", "Ajay")
register_student("S2", "pranav")

enroll_student("S1", "C101")
enroll_student("S2", "C101")
enroll_student("S2", "C102")
enroll_student("S1", "C102")

course_report("C101")
course_report("C102")
```
